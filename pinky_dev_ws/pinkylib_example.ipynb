```python
# pinky.buzzer()로 부저 울리는 횟수 지정가능

from pinkylib import Buzzer

pinky = Buzzer()
pinky.buzzer_start()
pinky.buzzer(3)
```

```python
# set_buzzer_duty() 함수로 부저의 소리 크기 설정

import time

pinky.set_buzzer_duty(50)
time.sleep(1)
pinky.set_buzzer_duty(0)
```

```python
# set_buzzer_freq()로 부저 높낮이 변경 가능

pinky.set_buzzer_freq(500)
pinky.set_buzzer_duty(50)
time.sleep(1)
pinky.set_buzzer_duty(0)
```

```python
# set_buzzer_freq(), set_buzzer_duty()로 조합으로 연주를 할 수 있다

# 미 (330Hz)
pinky.set_buzzer_freq(330)
pinky.set_buzzer_duty(50)
time.sleep(0.5)

# 레 (294Hz)
pinky.set_buzzer_freq(294)
pinky.set_buzzer_duty(50)
time.sleep(0.5)

# 도 (262Hz)
pinky.set_buzzer_freq(262)
pinky.set_buzzer_duty(50)
time.sleep(0.5)

# 레 (294Hz)
pinky.set_buzzer_freq(294)
pinky.set_buzzer_duty(50)
time.sleep(0.5)

# 부저 끄기
pinky.set_buzzer_duty(0)
pinky.close() # gpio 자원 해제
```

```python
from pinkylib import Ultrasonic
import time

pinky = Ultrasonic()

for i in range(5):
    dist = pinky.get_dist()
    print(f"{dist:.2f} m")
    time.sleep(0.5)
    
pinky.close()
```

```python
from pinkylib import IR
import time

pinky = IR()

for i in range(10):
    ir_1, ir_2, ir_3 = pinky.read_ir()
    print(f"ir_1: {ir_1}, ir_2: {ir_2}, ir_3: {ir_3}")
    time.sleep(0.5)
    
pinky.close()
```

```python
from pinkylib import IMU
import time

imu = IMU()

# 9축 모드 설정
#imu = IMU(mode='9dof') 

try:
    for i in range(10):
        imu_data = imu.read_imu_data()

        acc = imu_data['acc']
        mag = imu_data['mag']
        gyro = imu_data['gyro']
        euler = imu_data['euler']

        print("*"*10)
        print(f"ACC   | X:{acc[0]:6.2f} Y:{acc[1]:6.2f} Z:{acc[2]:6.2f} (m/s^2)")
        print(f"MAG   | X:{mag[0]:6.2f} Y:{mag[1]:6.2f} Z:{mag[2]:6.2f} (uT)")
        print(f"GYRO  | X:{gyro[0]:6.2f} Y:{gyro[1]:6.2f} Z:{gyro[2]:6.2f} (deg/s)")
        print(f"EULER | Roll:{euler[0]:6.2f} Pitch:{euler[1]:6.2f} Yaw:{euler[2]:6.2f} (deg)")

        time.sleep(0.5)
   
    imu.close()
except KeyboardInterrupt:
    print("Process stopped by user")
```
